```python
# Celda 0.1 - Verificacion del dataset de Reddit (solo lectura, no descarga contenido)
import os
from pathlib import Path
from collections import Counter

DATA = Path('/Users/ppizam/Library/CloudStorage/GoogleDrive-pizacapital@gmail.com'
            '/Other computers/My Mac RRG/data/reddit')
assert DATA.exists(), 'no se encuentra la carpeta data/reddit'

conteo = Counter()
ceros = []
subs_2026_sin_submissions = []

for anio in sorted(DATA.iterdir()):
    if not anio.is_dir() or not anio.name.isdigit():
        continue
    for mes in sorted(anio.iterdir()):
        if not mes.is_dir():
            continue
        archivos = [f for f in mes.iterdir() if f.suffix in ('.zst', '.ndjson')]
        for f in archivos:
            tipo = 'submissions' if 'submissions' in f.name else 'comments'
            conteo[(anio.name, tipo)] += 1
            if f.stat().st_size == 0:
                ceros.append(str(f.relative_to(DATA)))
        if anio.name == '2026':
            if not any('submissions' in f.name for f in archivos):
                subs_2026_sin_submissions.append(f'{anio.name}/{mes.name}')

print('Archivos por anio y tipo:')
for (a, t), n in sorted(conteo.items()):
    print(f'  {a} {t:12} {n}')
print('\nArchivos en CERO bytes:', len(ceros))
for c in ceros:
    print('  ', c)
print('\nMeses de 2026 sin submissions:', subs_2026_sin_submissions)
# Esperado: 2 ceros (Daytrading 2021/01 y 02) y los 6 meses de 2026 sin submissions
```
